## 1. Using `concat()` for Stacking Data
- `pd.concat()` combines DataFrames vertically (rows) or horizontally (columns).
- Use `axis=0` for vertical stacking (default), `axis=1` for horizontal.
- `ignore_index=True` can reset the index after concatenation.

```python
import pandas as pd
# Dataset 1: Product details
df1 = pd.DataFrame({
    'ProductID' : [101,102,103],
    'ProductName': ['Pen', 'Notebook', 'Marker'],
    'Price': [10,50,30]
})

# Dataset 2: More products (same columns as df1 → good for vertical stacking)
df2 = pd.DataFrame({
    'ProductID': [104, 105],
    'ProductName': ['Pencil', 'Eraser'],
    'Price': [5,4]
})
# Dataset 3: Additional info (different columns → good for horizontal stacking)
df3 = pd.DataFrame({
    'Stock': [100, 60, 80],
    'Rating': [4.5,4.7,4.8]
})

print(f'DF1 {df1}')
print(f'DF2 {df2}')
print(f'DF3 {df3}')

#Vertical stacking (axis=0)
vertical_stack = pd.concat([df1, df2], ignore_index=True)
print(f'Vertical Stack {vertical_stack}')

#2. Horizontal stacking (axis=1)
horizontal_stack = pd.concat([df1, df3], axis=1)
print(f'Horizontal Stack {horizontal_stack}')
```

## 2. Using `merge()` for SQL-Style Joins
- `merge()` is similar to SQL joins.
- Specify `on` or `left_on`/`right_on` to choose key columns.
- Join types: `inner` (default), `left`, `right`, `outer`.

```python
#you have to concat df1 and df 2 row wise
product = pd.concat([df1,df2], ignore_index=True)

sales = pd.DataFrame({
    'ProductID' : [101,102,104,106],
    'Units' : [500,200,150,50]
})
print(product)
print(sales)

#1. Inner join (Dono dataframes ke beech only matching ProductID)
inner = pd.merge(product, sales, how='inner')
print(inner)

#2. Left join
left = pd.merge(product, sales, how='left')
print(left)

#3. Right join
right = pd.merge(product, sales, how='right')
print(right)

#4. Outer join -> everthing, even unmatch
outer = pd.merge(product, sales, how='outer')
print(outer)
```

## 3. Using `join()` with Index Alignment
- `join()` joins DataFrames on their index or on a key column.
- By default, joins on index; can specify `on='col'`.
- Useful for adding columns to an existing DataFrame.

```python
sales = pd.DataFrame({
    'UnitsSold' : [500,200,150,50,350] ,
}, index=[101,102,103,104,105])
print(product)
print(sales)

#1. Join stock info to products .
p_indexed = product.set_index('ProductID')
print(p_indexed)
joined = p_indexed.join(sales)
print(joined)
```

## Mini Practice
1. Create three DataFrames: Products, Sales, and Revenue.
2. Concatenate them vertically and horizontally.
3. Merge Sales and Revenue using different join types.
4. Join Products with Sales based on index or a key.
5. Analyze missing values after joins.

```python

# 1. Create three DataFrames: Products, Sales, and Revenue
products = pd.DataFrame({
    "ProductID": [101, 102, 103, 104],
    "Product": ["Book", "Sketchbook", "Notebook", "SketchPen"]
})

sales = pd.DataFrame({
    "ProductID": [101, 102, 104, 106],
    "UnitsSold": [230, 100, 90, 40]
})

revenue = pd.DataFrame({
    "ProductID": [101, 102, 103, 105],
    "Revenue": [1000, 500, 450, 200]
})


print(products)
print(sales)
print(revenue)

```

```python
# 2. Concatenate them vertically and horizontally
vertical = pd.concat([sales, revenue], ignore_index=True)
print(vertical)

horizontal = pd.concat([products, sales], axis=1)
print(horizontal)


# 3. Merge Sales and Revenue using different join types (inner, left, right, outer)
inner = pd.merge(sales, revenue, on="ProductID", how="inner")
print(inner)

left = pd.merge(sales, revenue, on="ProductID", how="left")
print(left)

right = pd.merge(sales, revenue, on="ProductID", how="right")
print(right)

outer = pd.merge(sales, revenue, on="ProductID", how="outer")
print(outer)
```

```python
# 4. Join Products with Sales based on index or a key
products_index = products.set_index("ProductID")
sales_index = sales.set_index("ProductID")

joined = products_index.join(sales_index)

print(joined)

# 5. Analyze missing values after joins
# Use isna(), info(), or describe() to explore gaps
print(joined.isna())

print(joined.isna().sum())

joined.info()

print(joined.describe())
```
