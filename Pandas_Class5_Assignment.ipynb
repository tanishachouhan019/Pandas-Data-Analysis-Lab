## Q10. Sort DataFrame by Salary

Input:
Name   Salary
Aman   30000
Raj    50000
Riya   40000

Output:
Raj
Riya
Aman

```python
import pandas as pd
data = {
    'Name' : ['Aman','Raj','Riya'],
    'Salary' : [30000,50000,40000]
    }
df = pd.DataFrame(data)
Salary = df['Name']
print(Salary.sort_values())
```

## Q11. Find Correlation

Input:
Math = [10,20,30,40]
Science = [15,25,35,45]

Output:
1.0

```python
import pandas as pd
data = {
    'Math' : [10,20,30,40],
    'Science' : [15,25,35,45]
    }
df = pd.DataFrame(data)
correlation = df['Math'].corr(df['Science'])

print("correlation =",correlation)
```

## Q12. Find Covariance

Input:
Math = [10,20,30,40]
Science = [15,25,35,45]

Output:
166.67 (approx)

```python
import pandas as pd
data = {
    'Math' : [10,20,30,40],
    'Science' : [15,25,35,45]
    }
df = pd.DataFrame(data)
covariance = df['Math'].cov(df['Science'])

print("covarience =",covarience)
```

## Q13. Sort by Two Columns

Input:
Sort by City and then Age

Output:
Sorted DataFrame

```python
import pandas as pd
data = {
    'Age' : [20, 25, 25, 30, 25],
    'city' : ['Bhopal','Indore','Bhopal','Jabalpur','Raipur']
}
df = pd.DataFrame(data)

print(df.sort_values(by=['city','Age']))
```

## Q14. Student Statistics

Input:
Marks = [60,70,80,90]

Output:
Mean = 75
Median = 75
Max = 90
Min = 60

```python
Marks = [60,70,80,90]
print(f"Mean = {sum(Marks)/len(Marks)}")
print(f"Median = {sorted(Marks)[len(Marks)//2]}")
print(f"Max = {max(Marks)}")
print(f"Min = {min(Marks)}")
```

## Q15. Mini Practice

Create a DataFrame with:
Name, Age, Score, Height

Tasks:
1. Sort by Score (Descending)
2. Find Mean Age
3. Find Median Height
4. Show Score Frequency

```python
import pandas as pd
data = {
    'Name' : ['Aman','Raj','Riya'],
    'Age' : [20, 25, 30],
    'Score' : [30000,50000,40000],
    'Height' : [170,180,190]
    }
df = pd.DataFrame(data)
print(df.sort_values(by='Score',ascending=False))
print(f"Mean Age = {df['Age'].mean()}")
print(f"Median Height = {sorted(df['Height'])[len(df['Height'])//2]}")
print(df['Score'].value_counts())
```
