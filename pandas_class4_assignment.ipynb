# Q1. Select Single Column

## Problem Statement
Create a DataFrame and display only the Name column.

## Sample Input
```python
Name, Age
```

## Sample Output
```python
Only Name column
```


````python
import pandas as pd
a = {
    'Name' : ['Alice','Bob','Charlie'],
    'Age' : [25,30,22]
}
b = pd.DataFrame(a)
print(b)

print(b['Name'])
````


# Q2. Select Multiple Columns

## Problem Statement
Display Name and Marks columns from a DataFrame.

## Sample Input
```python
Name, Marks
```

## Sample Output
```python
Selected columns
```


````python
import pandas as pd
a = {
    'Name' : ['Alice','Bob','Charlie'],
    'Marks' : [90,85,77]
}
b = pd.DataFrame(a)
print(b)

print(b[['Name','Marks']])
````


# Q3. Filter Students

## Problem Statement
Display students with Marks greater than 80.

## Sample Input
```python
Marks > 80
```

## Sample Output
```python
Filtered rows
```


````python
print(b[b['Marks']>80])
````


# Q4. Filter by City

## Problem Statement
Display students belonging to Bhopal.

## Sample Input
```python
City = Bhopal
```

## Sample Output
```python
Matching rows
```


````python
import pandas as pd
a = {
    'Name' : ['Alice','Bob','Charlie'],
    'Marks' : [90,85,77],
    'City' : ['Bhopal','Indore','Bhopal']
}
b = pd.DataFrame(a)
print(b)

print(b[b['City']=='Bhopal'])
````


# Q5. Find Maximum Marks

```python
Marks column
```

## Sample Output

```python
Highest value
```


# Q6. Find Minimum Age

## Problem Statement
Find the minimum age.

## Sample Input
```python
Age column
```

## Sample Output
```python
Minimum age
```


````python
import pandas as pd
a = {
    'Name' : ['Alice','Bob','Charlie'],
    'Marks' : [90,85,77],
    'City' : ['Bhopal','Indore','Bhopal'],
    'Age' : [20,22,21]
}
b = pd.DataFrame(a)
print(b)

print(b['Age'].min())
````


# Q7. Sort Data


# Q9. Group By City

## Problem Statement
Find average marks city-wise.

## Sample Input
```python
City and Marks
```

## Sample Output
```python
Grouped result
```


````python
print(b.groupby('City')['Marks'].mean())
````
